```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Demo_Notebook").getOrCreate()
nums_rdd         = spark.sparkContext.parallelize([1, 2, 3, 4, 5, 6, 7, 3, 7, 2])
words_rdd        = spark.sparkContext.parallelize(["apple orange", "banana mango", "apple banana"])
kv_rdd1          = spark.sparkContext.parallelize([("a", 1), ("b", 2), ("a", 3)])
kv_rdd2          = spark.sparkContext.parallelize([("a", 10), ("c", 30)])
```

```python
# Squares of Numbers
Squares = nums_rdd.map(lambda x: x**2)
print("Map -->", Squares.collect())
```

```python
# flatMap(flattens the result): split each sentence into words (flattens)
words = words_rdd.flatMap(lambda line: line.split(" "))
print("FlatMap-->", words.collect())
```

```python
# Filter: Filter the even
even = nums_rdd.filter(lambda x: x%2 == 0)
print("Even numbers:", even.collect())
```

```python
# Distinct: Removes the Duplicates
Distinct_values = nums_rdd.distinct()
Distinct_words = words.distinct()
print("Distinct numbers:", Distinct_values.collect())
print("Distinct words:", Distinct_words.collect())
```

```python
# Union: Combines two RDDs
union_values = nums_rdd.union(spark.sparkContext.parallelize([9, 10, 11]))
print("union of two RDDs:", union_values.collect())
```

```python
# Intersection: gets the common elements
intersect_values = nums_rdd.intersection(spark.sparkContext.parallelize([4,6,8]))
print("Interesect values:", intersect_values.collect())
```

```python
# Subtraction: like A minus B
Subtracted_values = nums_rdd.subtract(spark.sparkContext.parallelize([1,2,3,4]))
print("Subtracted_values:", Subtracted_values.collect())
```

```python
 # Cartesian: makes the ordered pairs
 cart = spark.sparkContext.parallelize([1, 2]).cartesian(spark.sparkContext.parallelize(["a", "b"]))
 print("Cartesian pairs:", cart.collect())
```
